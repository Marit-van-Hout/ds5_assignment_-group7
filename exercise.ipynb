```python

# Regression coefficients from your model
intercept = 20520
beta_years = -201.476
beta_km1000 = -42.5092

# Target car: 5 years old, 80,000 km → km1000 = 80
years = 5
km1000 = 80
```

```python
# a. Expected price
expected_price = intercept + beta_years * years + beta_km1000 * km1000
print("Expected price:", round(expected_price, 2))
```

```python
# b. Multicollinearity check (based on correlation matrix)
# years & km1000 have correlation 0.977 → multicollinear
multicollinear_variables = ["years", "km1000"]
print("Multicollinear variables:", multicollinear_variables)
```

```python
# c. Interpretation of beta_years
interpretation_years = (
    "Each additional year reduces the price by approximately €201.48, "
    "holding kilometers constant."
)
print("Interpretation β1 (years):", interpretation_years)
```

```python
# d. Test linear relationship between price and km1000
t_value_km = -5.842
p_value_km = "< 0.001"  # extremely significant
print("Linear relationship price ~ km1000: significant (t =", t_value_km, ")")
```

```python
# e. R² adjusted, F-statistic, p-value
R2 = 0.998
n = 5
k = 2

R2_adj = 1 - (1 - R2) * ((n - 1) / (n - k - 1))
print("Adjusted R²:", round(R2_adj, 3))

F_stat = (R2 / k) / ((1 - R2) / (n - k - 1))
print("F-statistic:", round(F_stat, 1))

p_value_F = "< 0.001"  # extremely small
print("P-value (F-statistic):", p_value_F)
```
